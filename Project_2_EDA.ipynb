```python
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files

uploaded = files.upload()

file_name = list(uploaded.keys())[0]
df = pd.read_excel(file_name)

df.head()
```

```python
# Dataset ki basic information
df.info()

# Pehli 5 rows
df.head()

# Statistical summary
df.describe(include='all')
```

```python
# Mean, Median aur Count (sirf numerical columns ke liye)

print("Mean:")
print(df.mean(numeric_only=True))

print("\nMedian:")
print(df.median(numeric_only=True))

print("\nCount:")
print(df.count())
```

```python
import matplotlib.pyplot as plt

# Histogram - Total Price
plt.figure(figsize=(8,5))
plt.hist(df["TotalPrice"], bins=20)
plt.title("Distribution of Total Price")
plt.xlabel("Total Price")
plt.ylabel("Frequency")
plt.show()

# Bar Chart - Product Count
plt.figure(figsize=(8,5))
df["Product"].value_counts().plot(kind="bar")
plt.title("Product Distribution")
plt.xlabel("Product")
plt.ylabel("Count")
plt.show()

# Box Plot - Total Price (Outliers)
plt.figure(figsize=(6,4))
plt.boxplot(df["TotalPrice"])
plt.title("Box Plot of Total Price")
plt.show()
```

```python
print("Key Observations:")
print("1. Dataset contains 1200 records.")
print("2. CouponCode column has missing values.")
print("3. Average Quantity:", round(df["Quantity"].mean(),2))
print("4. Average Total Price:", round(df["TotalPrice"].mean(),2))
print("5. Median Total Price:", round(df["TotalPrice"].median(),2))
```
