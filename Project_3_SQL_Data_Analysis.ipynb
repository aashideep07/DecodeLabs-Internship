```python
import pandas as pd

df = pd.read_excel("Dataset for Data Analytics.xlsx")

df.head()
```

```python
import sqlite3

# SQLite database create
conn = sqlite3.connect("project3.db")

# DataFrame ko SQL table me save karo
df.to_sql("orders", conn, if_exists="replace", index=False)

print("Data successfully loaded into SQL table!")
```

```python
query = """
SELECT * FROM orders
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT * FROM orders
WHERE OrderStatus = 'Delivered';
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT * FROM orders
ORDER BY TotalPrice DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT Product, COUNT(*) AS Total_Orders
FROM orders
GROUP BY Product
ORDER BY Total_Orders DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT PaymentMethod, SUM(TotalPrice) AS Total_Sales
FROM orders
GROUP BY PaymentMethod;===
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT AVG(Quantity) AS Average_Quantity
FROM orders;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT OrderID, Product, TotalPrice
FROM orders
ORDER BY TotalPrice DESC
LIMIT 5;
"""

pd.read_sql(query, conn)
```
